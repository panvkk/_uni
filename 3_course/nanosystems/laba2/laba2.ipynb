## Задание 1

```python
import numpy as np


def create_matrix(n):
    # Создает матрицу размера n x n без циклов
    diagonal = -np.arange(n, 0, -1, dtype=float)
    matrix = np.diag(diagonal)
    matrix += np.diag(np.full(n - 1, 4.0), k=1)
    matrix[3, :] = -4.0
    return matrix


def save_matrix(matrix, filename="matrix_A.txt"):
    # Сохраняет матрицу в текстовый файл 
    np.savetxt(filename, matrix, fmt="%.6g")


N = 10
A = create_matrix(N)
save_matrix(A)

print("Матрица A:")
print(A)
print("Матрица сохранена в файл matrix_A.txt")
```

## Задание 2

```python
def octahedral_norm(matrix):
    # Октаэдрическая (1-)норма матрицы или вектора"""
    return np.linalg.norm(matrix, ord=1)


def analyze_matrix(filename="matrix_A.txt"):
    # Читает матрицу, анализирует обратную матрицу и решает Ax=b
    matrix = np.loadtxt(filename)
    vector = np.ones(matrix.shape[0])

    if np.isclose(np.linalg.det(matrix), 0.0):
        print("Матрица A вырождена: обратная матрица и решение не определены.")
        return None

    inverse_matrix = np.linalg.inv(matrix)
    solution = np.linalg.solve(matrix, vector)

    print("Октаэдрическая норма матрицы A^(-1):", octahedral_norm(inverse_matrix))
    print("Решение системы Ax = b:")
    print(solution)
    print("Октаэдрическая норма решения:", octahedral_norm(solution))
    return inverse_matrix, solution


analysis_result = analyze_matrix()
```

## Задание 3

Первая функция использует цикл, а вторая вычисляет сумму без циклов с помощью векторных операций NumPy. Затем сравнивается время выполнения функций.

```python
import timeit


def sum_with_loop(z, n=10):
    # Вычисляет сумму с использованием цикла
    total = 0.0j if np.iscomplexobj(z) else 0.0
    for k in range(1, n + 1):
        total += np.arctan(z**3 / (k * np.sqrt(k)))
    return total


def sum_with_numpy(z, n=10):
    # Вычисляет сумму без циклов
    k = np.arange(1, n + 1, dtype=float)
    terms = np.arctan(z**3 / (k * np.sqrt(k)))
    return np.sum(terms)


z = 1 + 1j
N_sum = 10

sum_loop = sum_with_loop(z, N_sum)
sum_numpy = sum_with_numpy(z, N_sum)

print("Результат функции с циклом:", sum_loop)
print("Результат функции без циклов:", sum_numpy)
print("Результаты совпадают:", np.allclose(sum_loop, sum_numpy))

repetitions = 10000
time_loop = timeit.timeit(lambda: sum_with_loop(z, N_sum), number=repetitions)
time_numpy = timeit.timeit(lambda: sum_with_numpy(z, N_sum), number=repetitions)

print(f"Время функции с циклом ({repetitions} запусков): {time_loop:.6f} с")
print(f"Время функции без циклов ({repetitions} запусков): {time_numpy:.6f} с")
```
