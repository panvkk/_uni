```python
import math
import random


def circles_intersect(circle_1, circle_2):
    # Возвращает True, если окружности имеют хотя бы одну общую точку.
    x_1, y_1, radius_1 = circle_1
    x_2, y_2, radius_2 = circle_2

    if radius_1 < 0 or radius_2 < 0:
        raise ValueError("Радиус окружности не может быть отрицательным")

    distance = math.hypot(x_2 - x_1, y_2 - y_1)
    return abs(radius_1 - radius_2) <= distance <= radius_1 + radius_2


def swap_first_min_max(numbers):
    # Меняет местами первое вхождение минимума и первое вхождение максимума.
    if not numbers:
        raise ValueError("Список не должен быть пустым")

    result = numbers.copy()
    minimum_index = result.index(min(result))
    maximum_index = result.index(max(result))
    result[minimum_index], result[maximum_index] = (
        result[maximum_index],
        result[minimum_index],
    )
    return result
```

## Задание 2

Вводятся шесть чисел: координаты центра и радиус первой окружности `(x1, y1, R1)`, затем координаты центра и радиус второй окружности `(x2, y2, R2)`.

Расстояние между центрами:

`d = sqrt((x2 - x1)^2 + (y2 - y1)^2)`.

Окружности имеют хотя бы одну общую точку, если выполняется условие:

`|R1 - R2| <= d <= R1 + R2`.

Это условие учитывает пересечение и внешнее или внутреннее касание окружностей.

```python
print("Введите данные первой окружности:")
x_1, y_1, radius_1 = map(float, input("x1 y1 R1: ").split())

print("Введите данные второй окружности:")
x_2, y_2, radius_2 = map(float, input("x2 y2 R2: ").split())

result = circles_intersect(
    (x_1, y_1, radius_1),
    (x_2, y_2, radius_2),
)

print("Окружности пересекаются:", result)
```

## Задание 3

Список формируется случайным образом. После этого первый минимальный и первый максимальный элементы меняются местами.

```python
list_size = int(input("Введите размер списка: "))
left_bound = int(input("Введите левую границу диапазона: "))
right_bound = int(input("Введите правую границу диапазона: "))

source_list = [
    random.randint(left_bound, right_bound)
    for _ in range(list_size)
]
changed_list = swap_first_min_max(source_list)

print("Исходный список:", source_list)
print("Список после обмена:", changed_list)
print("Индекс первого минимума:", source_list.index(min(source_list)))
print("Индекс первого максимума:", source_list.index(max(source_list)))
```
